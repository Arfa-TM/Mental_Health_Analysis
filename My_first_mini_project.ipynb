Developed a menu-based billing system for a tea shop using Python.

Simulated a real-time order system that displays the menu, takes multiple orders, and outputs the final bill.

```python
#Define the menu of the Tea Shop
```

```python
menu={'Tea': 15, 'Masala tea' : 20, 'Chocolate tea': 20, 'Lemon tea': 15, 'Honey tea' : 15}
```

```python
print(menu)
```

```python
#Greeting customers
```

```python
print('Welcome to My Tea Shop')
print('Tea:Rs.15\nMasala tea:Rs.20\nChocolate tea:Rs.20\nLemon tea:Rs.15\nHoney tea:Rs.15')

Order_total=0

item_1=input("Enter the Tea you would like to have = ")
if item_1 in menu:
    Order_total+=menu[item_1]
    print(f"Your first order {item_1} has been placed")
    
else:
    print(f"Sorry, we regret for the inconvinience")

Another_order=input("Do you want to have anything else? (Yes/No) ")
if Another_order =="Yes":
    item_2=input("Enter the name of the Tea you would like to have next =")
    if item_2 in menu:
        Order_total+=menu[item_2]
        print(f"Your order has been updated with {item_2}")
    else:
        print(f"Sorry, we regret for the inconvinience!")
print(f"Your total bill is {Order_total}")
```
